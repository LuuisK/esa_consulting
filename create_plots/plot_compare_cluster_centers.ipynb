```python
# ============================================================
# Vergleich der Clusterzentren von GMM und k-means
#
# Discovery:  GMM Discovery  vs. k-means Discovery
# Validation: GMM Validation vs. k-means Validation
#
# Datenbasis: MFA-Dimensionen
# Distanzmaß: euklidische Distanz
# Zuordnung: jeweils kleinstmögliche Distanz
# ============================================================

import sys
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from IPython.display import display
from scipy.spatial.distance import cdist


# ============================================================
# 0. Projektpfad
# ============================================================

project_root = Path.cwd().parent

if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from utils.saving_utils import DatasetBundle


# ============================================================
# 1. MFA-Daten laden
# ============================================================

data_discov_mfa = pd.read_parquet(
    project_root
    / "data"
    / "dim_red"
    / "data_discov_mfa.parquet"
).reset_index(drop=True)

data_valid_mfa = pd.read_parquet(
    project_root
    / "data"
    / "dim_red"
    / "data_valid_mfa.parquet"
).reset_index(drop=True)


# MFA-Dimensionen numerisch konvertieren
data_discov_mfa = data_discov_mfa.apply(
    pd.to_numeric,
    errors="coerce"
)

data_valid_mfa = data_valid_mfa.apply(
    pd.to_numeric,
    errors="coerce"
)


# ============================================================
# 2. Surveygewichte laden
# ============================================================

data_discov_full = DatasetBundle.load(
    project_root
    / "data"
    / "preprocessed"
    / "data_discov_full"
)

data_valid_full = DatasetBundle.load(
    project_root
    / "data"
    / "preprocessed"
    / "data_valid_full"
)

sample_weight_discov = pd.to_numeric(
    data_discov_full.data["wichtn"],
    errors="coerce"
).to_numpy(dtype=float)

sample_weight_valid = pd.to_numeric(
    data_valid_full.data["wichtn"],
    errors="coerce"
).to_numpy(dtype=float)


# ============================================================
# 3. Gespeicherte GMM-Ergebnisse laden
# ============================================================

gmm_discov_predictions = pd.read_parquet(
    project_root
    / "outputs"
    / "gmm_clustering"
    / "cluster_pred_discov.parquet"
).reset_index(drop=True)

gmm_valid_predictions = pd.read_parquet(
    project_root
    / "outputs"
    / "gmm_clustering"
    / "cluster_pred_valid.parquet"
).reset_index(drop=True)


# ============================================================
# 4. Gespeicherte k-means-Ergebnisse laden
# ============================================================

kmeans_discov_predictions = pd.read_parquet(
    project_root
    / "outputs"
    / "kmeans_clustering"
    / "cluster_pred_discov.parquet"
).reset_index(drop=True)

kmeans_valid_predictions = pd.read_parquet(
    project_root
    / "outputs"
    / "kmeans_clustering"
    / "cluster_pred_valid.parquet"
).reset_index(drop=True)


# ============================================================
# 5. Plausibilitätsprüfungen
# ============================================================

def check_inputs(
    mfa_data: pd.DataFrame,
    weights: np.ndarray,
    gmm_predictions: pd.DataFrame,
    kmeans_predictions: pd.DataFrame,
    sample_name: str
) -> None:
    """
    Prüft, ob MFA-Daten, Gewichte und Clusterergebnisse
    dieselbe Anzahl an Personen enthalten.
    """
    lengths = {
        "MFA-Daten": len(mfa_data),
        "Gewichte": len(weights),
        "GMM-Ergebnisse": len(gmm_predictions),
        "k-means-Ergebnisse": len(kmeans_predictions),
    }

    if len(set(lengths.values())) != 1:
        raise ValueError(
            f"{sample_name}: unterschiedliche Längen:\n"
            f"{lengths}"
        )

    if mfa_data.isna().any().any():
        missing_columns = (
            mfa_data.columns[
                mfa_data.isna().any()
            ]
            .tolist()
        )

        raise ValueError(
            f"{sample_name}: Die MFA-Daten enthalten "
            f"fehlende Werte in: {missing_columns}"
        )

    if not np.isfinite(weights).any():
        raise ValueError(
            f"{sample_name}: Es liegen keine gültigen "
            f"Surveygewichte vor."
        )


check_inputs(
    data_discov_mfa,
    sample_weight_discov,
    gmm_discov_predictions,
    kmeans_discov_predictions,
    "Discovery"
)

check_inputs(
    data_valid_mfa,
    sample_weight_valid,
    gmm_valid_predictions,
    kmeans_valid_predictions,
    "Validation"
)

print("Discovery-MFA:", data_discov_mfa.shape)
print("Validation-MFA:", data_valid_mfa.shape)


# ============================================================
# 6. GMM-Zentren im MFA-Raum berechnen
# ============================================================

def calculate_gmm_centers(
    mfa_data: pd.DataFrame,
    predictions: pd.DataFrame,
    survey_weights: np.ndarray
) -> pd.DataFrame:
    """
    Berechnet für jede GMM-Komponente den Mittelpunkt im
    MFA-Raum.

    Verwendetes Gewicht pro Person und Komponente:

        Surveygewicht × posteriore Zugehörigkeitswahrscheinlichkeit

    Dadurch werden die weichen GMM-Zuordnungen berücksichtigt.
    """

    probability_columns = [
        column
        for column in predictions.columns
        if column != "label"
    ]

    if not probability_columns:
        raise ValueError(
            "Die GMM-Datei enthält keine posterioren "
            "Zugehörigkeitswahrscheinlichkeiten."
        )

    probabilities = predictions[
        probability_columns
    ].apply(
        pd.to_numeric,
        errors="coerce"
    )

    if probabilities.isna().any().any():
        raise ValueError(
            "Die GMM-Wahrscheinlichkeiten enthalten "
            "fehlende oder nicht numerische Werte."
        )

    survey_weights = np.asarray(
        survey_weights,
        dtype=float
    )

    centers = {}

    for component_number, column in enumerate(
        probability_columns
    ):
        posterior = probabilities[
            column
        ].to_numpy(dtype=float)

        combined_weights = (
            survey_weights
            * posterior
        )

        usable = (
            np.isfinite(combined_weights)
            & (combined_weights > 0)
        )

        if usable.sum() == 0:
            raise ValueError(
                f"Für GMM-Komponente {component_number} "
                f"liegen keine gültigen Gewichte vor."
            )

        centers[component_number] = np.average(
            mfa_data.loc[
                usable
            ].to_numpy(dtype=float),
            axis=0,
            weights=combined_weights[usable]
        )

    result = pd.DataFrame.from_dict(
        centers,
        orient="index",
        columns=mfa_data.columns
    )

    result.index.name = "gmm_cluster"

    return result


# ============================================================
# 7. k-means-Zentren im MFA-Raum berechnen
# ============================================================

def calculate_kmeans_centers(
    mfa_data: pd.DataFrame,
    predictions: pd.DataFrame,
    survey_weights: np.ndarray
) -> pd.DataFrame:
    """
    Berechnet für jedes k-means-Cluster den gewichteten
    Mittelpunkt im MFA-Raum anhand der harten Zuordnungen.
    """

    labels = pd.to_numeric(
        predictions["label"],
        errors="raise"
    ).astype(int)

    survey_weights = np.asarray(
        survey_weights,
        dtype=float
    )

    centers = {}

    for cluster in sorted(
        labels.unique()
    ):
        cluster_mask = (
            labels.to_numpy()
            == cluster
        )

        cluster_weights = survey_weights[
            cluster_mask
        ]

        cluster_data = (
            mfa_data.loc[
                cluster_mask
            ]
            .reset_index(drop=True)
        )

        usable = (
            np.isfinite(cluster_weights)
            & (cluster_weights > 0)
        )

        if usable.sum() == 0:
            raise ValueError(
                f"Für k-means-Cluster {cluster} "
                f"liegen keine gültigen Gewichte vor."
            )

        centers[cluster] = np.average(
            cluster_data.loc[
                usable
            ].to_numpy(dtype=float),
            axis=0,
            weights=cluster_weights[usable]
        )

    result = pd.DataFrame.from_dict(
        centers,
        orient="index",
        columns=mfa_data.columns
    )

    result.index.name = "kmeans_cluster"

    return result


# ============================================================
# 8. Zentren für beide Wellen berechnen
# ============================================================

gmm_discov_centers = calculate_gmm_centers(
    data_discov_mfa,
    gmm_discov_predictions,
    sample_weight_discov
)

kmeans_discov_centers = calculate_kmeans_centers(
    data_discov_mfa,
    kmeans_discov_predictions,
    sample_weight_discov
)

gmm_valid_centers = calculate_gmm_centers(
    data_valid_mfa,
    gmm_valid_predictions,
    sample_weight_valid
)

kmeans_valid_centers = calculate_kmeans_centers(
    data_valid_mfa,
    kmeans_valid_predictions,
    sample_weight_valid
)


print("\nAnzahl der Clusterzentren:")

print(
    "Discovery:",
    f"GMM={len(gmm_discov_centers)},",
    f"k-means={len(kmeans_discov_centers)}"
)

print(
    "Validation:",
    f"GMM={len(gmm_valid_centers)},",
    f"k-means={len(kmeans_valid_centers)}"
)


# ============================================================
# 9. Kleinste Distanz statt Hungarian Matching
# ============================================================

def compare_cluster_centers(
    gmm_centers: pd.DataFrame,
    kmeans_centers: pd.DataFrame,
    sample_name: str
) -> dict:
    """
    Vergleicht GMM- und k-means-Zentren innerhalb einer Welle.

    Für jedes GMM-Cluster wird das k-means-Cluster mit der
    kleinsten euklidischen Distanz bestimmt.

    Zusätzlich wird umgekehrt für jedes k-means-Cluster das
    nächstgelegene GMM-Cluster bestimmt.

    Es gibt keine One-to-one-Bedingung. Mehrere GMM-Cluster
    dürfen daher demselben k-means-Cluster zugeordnet werden.
    """

    if list(gmm_centers.columns) != list(
        kmeans_centers.columns
    ):
        raise ValueError(
            f"{sample_name}: GMM und k-means verwenden "
            f"nicht dieselben MFA-Dimensionen."
        )

    # --------------------------------------------------------
    # Vollständige euklidische Distanzmatrix
    # --------------------------------------------------------

    distances = cdist(
        gmm_centers.to_numpy(dtype=float),
        kmeans_centers.to_numpy(dtype=float),
        metric="euclidean"
    )

    distance_matrix = pd.DataFrame(
        distances,
        index=gmm_centers.index,
        columns=kmeans_centers.index
    )

    all_distances = distances.ravel()

    overall_mean = float(
        np.mean(all_distances)
    )

    overall_median = float(
        np.median(all_distances)
    )

    # --------------------------------------------------------
    # Für jedes GMM-Cluster:
    # nächstes k-means-Cluster
    # --------------------------------------------------------

    nearest_kmeans_positions = np.argmin(
        distances,
        axis=1
    )

    nearest_kmeans_distances = np.min(
        distances,
        axis=1
    )

    nearest_kmeans_labels = (
        distance_matrix.columns[
            nearest_kmeans_positions
        ]
    )

    gmm_to_kmeans = pd.DataFrame({
        "gmm_cluster": distance_matrix.index,
        "nearest_kmeans_cluster": (
            nearest_kmeans_labels
        ),
        "minimum_distance": (
            nearest_kmeans_distances
        )
    })

    # Distanz relativ zur typischen Distanz
    gmm_to_kmeans[
        "relative_to_median_distance"
    ] = (
        gmm_to_kmeans["minimum_distance"]
        / overall_median
    )

    # Anteil aller Distanzen, die kleiner oder gleich sind
    gmm_to_kmeans[
        "percentile_among_all_distances"
    ] = [
        100 * np.mean(
            all_distances <= distance
        )
        for distance
        in gmm_to_kmeans["minimum_distance"]
    ]

    # Sortierung:
    # erst nach nächstem k-means-Cluster,
    # dann nach der Distanz
    gmm_to_kmeans = (
        gmm_to_kmeans
        .sort_values(
            [
                "nearest_kmeans_cluster",
                "minimum_distance"
            ],
            ascending=[True, True]
        )
        .reset_index(drop=True)
    )

    # --------------------------------------------------------
    # Umgekehrt:
    # Für jedes k-means-Cluster nächstes GMM-Cluster
    # --------------------------------------------------------

    nearest_gmm_positions = np.argmin(
        distances,
        axis=0
    )

    nearest_gmm_distances = np.min(
        distances,
        axis=0
    )

    nearest_gmm_labels = (
        distance_matrix.index[
            nearest_gmm_positions
        ]
    )

    kmeans_to_gmm = pd.DataFrame({
        "kmeans_cluster": distance_matrix.columns,
        "nearest_gmm_cluster": (
            nearest_gmm_labels
        ),
        "minimum_distance": (
            nearest_gmm_distances
        )
    })

    kmeans_to_gmm[
        "relative_to_median_distance"
    ] = (
        kmeans_to_gmm["minimum_distance"]
        / overall_median
    )

    kmeans_to_gmm[
        "percentile_among_all_distances"
    ] = [
        100 * np.mean(
            all_distances <= distance
        )
        for distance
        in kmeans_to_gmm["minimum_distance"]
    ]

    kmeans_to_gmm = (
        kmeans_to_gmm
        .sort_values(
            "minimum_distance",
            ascending=True
        )
        .reset_index(drop=True)
    )

    # --------------------------------------------------------
    # Gegenseitig nächste Nachbarn bestimmen
    # --------------------------------------------------------
    #
    # Ein Paar ist ein gegenseitiger nächster Nachbar, wenn:
    #
    # GMM g -> nächstes k-means k
    # und
    # k-means k -> nächstes GMM g
    #

    nearest_gmm_for_kmeans = {
        row["kmeans_cluster"]:
            row["nearest_gmm_cluster"]
        for _, row in kmeans_to_gmm.iterrows()
    }

    gmm_to_kmeans[
        "mutual_nearest_neighbors"
    ] = gmm_to_kmeans.apply(
        lambda row: (
            nearest_gmm_for_kmeans[
                row["nearest_kmeans_cluster"]
            ]
            == row["gmm_cluster"]
        ),
        axis=1
    )

    # --------------------------------------------------------
    # Heatmap anordnen
    # --------------------------------------------------------

    # Reihenfolge der GMM-Cluster entsprechend der Tabelle:
    # Cluster mit demselben nächsten k-means-Zentrum stehen
    # direkt untereinander.
    ordered_gmm_clusters = (
        gmm_to_kmeans["gmm_cluster"]
        .tolist()
    )

    ordered_kmeans_clusters = sorted(
        distance_matrix.columns
    )

    distance_matrix_ordered = (
        distance_matrix.loc[
            ordered_gmm_clusters,
            ordered_kmeans_clusters
        ]
    )

    # --------------------------------------------------------
    # Ergebnisse ausgeben
    # --------------------------------------------------------

    print("\n" + "=" * 72)
    print(
        f"GMM versus k-means cluster centres: "
        f"{sample_name}"
    )
    print("=" * 72)

    print(
        "\nFür jedes GMM-Cluster das nächstgelegene "
        "k-means-Cluster:"
    )

    display(
        gmm_to_kmeans.round(3)
    )

    print(
        "\nFür jedes k-means-Cluster das nächstgelegene "
        "GMM-Cluster:"
    )

    display(
        kmeans_to_gmm.round(3)
    )

    print("\nZusammenfassung der minimalen GMM-Distanzen:")

    display(
        gmm_to_kmeans[
            "minimum_distance"
        ]
        .describe()
        .round(3)
        .to_frame()
    )

    print("\nZusammenfassung aller paarweisen Distanzen:")

    display(
        pd.Series(
            all_distances,
            name="euclidean_distance"
        )
        .describe()
        .round(3)
        .to_frame()
    )

    print(
        "\nMittlere minimale Distanz "
        "(GMM → k-means):",
        round(
            gmm_to_kmeans[
                "minimum_distance"
            ].mean(),
            3
        )
    )

    print(
        "Mediane minimale Distanz "
        "(GMM → k-means):",
        round(
            gmm_to_kmeans[
                "minimum_distance"
            ].median(),
            3
        )
    )

    print(
        "Median aller paarweisen Distanzen:",
        round(
            overall_median,
            3
        )
    )

    print(
        "Mittelwert aller paarweisen Distanzen:",
        round(
            overall_mean,
            3
        )
    )

    print(
        "Anzahl gegenseitiger nächster Nachbarn:",
        int(
            gmm_to_kmeans[
                "mutual_nearest_neighbors"
            ].sum()
        )
    )

    # Wie viele GMM-Cluster werden jeweils einem
    # k-means-Cluster zugeordnet?
    gmm_clusters_per_kmeans = (
        gmm_to_kmeans
        .groupby(
            "nearest_kmeans_cluster"
        )
        .agg(
            number_of_gmm_clusters=(
                "gmm_cluster",
                "size"
            ),
            mean_minimum_distance=(
                "minimum_distance",
                "mean"
            ),
            median_minimum_distance=(
                "minimum_distance",
                "median"
            ),
            closest_gmm_distance=(
                "minimum_distance",
                "min"
            ),
            furthest_gmm_distance=(
                "minimum_distance",
                "max"
            )
        )
        .sort_index()
    )

    print(
        "\nAnzahl der GMM-Cluster pro nächstem "
        "k-means-Cluster:"
    )

    display(
        gmm_clusters_per_kmeans.round(3)
    )

    # --------------------------------------------------------
    # Heatmap
    # --------------------------------------------------------

    figure_width = max(
        8,
        len(
            distance_matrix_ordered.columns
        ) * 0.8
    )

    figure_height = max(
        7,
        len(
            distance_matrix_ordered.index
        ) * 0.42
    )

    plt.figure(
        figsize=(
            figure_width,
            figure_height
        )
    )

    image = plt.imshow(
        distance_matrix_ordered.to_numpy(
            dtype=float
        ),
        aspect="auto"
    )

    plt.colorbar(
        image,
        label="Euclidean distance"
    )

    plt.xticks(
        ticks=np.arange(
            len(
                distance_matrix_ordered.columns
            )
        ),
        labels=[
            f"K {cluster}"
            for cluster
            in distance_matrix_ordered.columns
        ],
        rotation=90
    )

    plt.yticks(
        ticks=np.arange(
            len(
                distance_matrix_ordered.index
            )
        ),
        labels=[
            f"G {cluster}"
            for cluster
            in distance_matrix_ordered.index
        ]
    )

    plt.xlabel("k-means clusters")
    plt.ylabel("GMM clusters")

    plt.title(
        f"GMM and k-means centre distances: "
        f"{sample_name}\n"
        "GMM clusters grouped by their nearest "
        "k-means centre"
    )

    plt.tight_layout()
    plt.show()

    return {
        "distance_matrix": distance_matrix,
        "distance_matrix_ordered": (
            distance_matrix_ordered
        ),
        "gmm_to_kmeans": gmm_to_kmeans,
        "kmeans_to_gmm": kmeans_to_gmm,
        "gmm_clusters_per_kmeans": (
            gmm_clusters_per_kmeans
        ),
        "median_all_distances": overall_median,
        "mean_all_distances": overall_mean
    }


# ============================================================
# 10. Discovery: GMM versus k-means
# ============================================================

results_discov_centers = compare_cluster_centers(
    gmm_centers=gmm_discov_centers,
    kmeans_centers=kmeans_discov_centers,
    sample_name="Discovery"
)


# ============================================================
# 11. Validation: GMM versus k-means
# ============================================================

results_valid_centers = compare_cluster_centers(
    gmm_centers=gmm_valid_centers,
    kmeans_centers=kmeans_valid_centers,
    sample_name="Validation"
)
```

```python
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401


def plot_centers_3d(
    gmm_centers: pd.DataFrame,
    kmeans_centers: pd.DataFrame,
    gmm_to_kmeans: pd.DataFrame,
    sample_name: str
) -> None:
    """
    Visualisiert GMM- und k-means-Zentren anhand der ersten
    drei MFA-Dimensionen.

    Linien verbinden jedes GMM-Zentrum mit seinem jeweils
    nächstgelegenen k-means-Zentrum.
    """

    if gmm_centers.shape[1] < 3 or kmeans_centers.shape[1] < 3:
        raise ValueError(
            "Für eine 3D-Darstellung werden mindestens "
            "drei MFA-Dimensionen benötigt."
        )

    dimensions = gmm_centers.columns[:3]

    fig = plt.figure(figsize=(11, 8))
    ax = fig.add_subplot(111, projection="3d")

    # GMM-Zentren
    ax.scatter(
        gmm_centers[dimensions[0]],
        gmm_centers[dimensions[1]],
        gmm_centers[dimensions[2]],
        marker="o",
        s=55,
        label="GMM centres"
    )

    # k-means-Zentren
    ax.scatter(
        kmeans_centers[dimensions[0]],
        kmeans_centers[dimensions[1]],
        kmeans_centers[dimensions[2]],
        marker="^",
        s=90,
        label="k-means centres"
    )

    # Clusterlabels
    for cluster, row in gmm_centers.iterrows():
        ax.text(
            row[dimensions[0]],
            row[dimensions[1]],
            row[dimensions[2]],
            f"G{cluster}",
            fontsize=8
        )

    for cluster, row in kmeans_centers.iterrows():
        ax.text(
            row[dimensions[0]],
            row[dimensions[1]],
            row[dimensions[2]],
            f"K{cluster}",
            fontsize=9
        )

    # GMM-Zentrum mit nächstem k-means-Zentrum verbinden
    for _, match in gmm_to_kmeans.iterrows():
        gmm_cluster = match["gmm_cluster"]
        kmeans_cluster = match["nearest_kmeans_cluster"]

        g = gmm_centers.loc[gmm_cluster, dimensions]
        k = kmeans_centers.loc[kmeans_cluster, dimensions]

        ax.plot(
            [g.iloc[0], k.iloc[0]],
            [g.iloc[1], k.iloc[1]],
            [g.iloc[2], k.iloc[2]],
            linewidth=0.8,
            alpha=0.5
        )

    ax.set_xlabel(str(dimensions[0]))
    ax.set_ylabel(str(dimensions[1]))
    ax.set_zlabel(str(dimensions[2]))

    ax.set_title(
        f"GMM and k-means cluster centres: {sample_name}\n"
        "Projection onto the first three MFA dimensions"
    )

    ax.legend()
    plt.tight_layout()
    plt.show()
```

```python
plot_centers_3d(
    gmm_centers=gmm_discov_centers,
    kmeans_centers=kmeans_discov_centers,
    gmm_to_kmeans=results_discov_centers["gmm_to_kmeans"],
    sample_name="Discovery"
)

plot_centers_3d(
    gmm_centers=gmm_valid_centers,
    kmeans_centers=kmeans_valid_centers,
    gmm_to_kmeans=results_valid_centers["gmm_to_kmeans"],
    sample_name="Validation"
)
```
