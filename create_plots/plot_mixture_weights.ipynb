```python
# Update path and working directory with project root
import os
import sys
from pathlib import Path

project_path = next(path for path in [Path.cwd(), *Path.cwd().parents] if (path / "pyproject.toml").is_file())
os.chdir(project_path)
sys.path.insert(0, str(project_path))
```

```python
# Imports
from pathlib import Path

import pandas as pd
import plotnine as p9

from utils.plotting_utils import make_plot_mixture_weights
from utils.saving_utils import load_object, save_plot

pd.options.mode.copy_on_write = True
pd.options.future.infer_string = True
```

```python
# Load data
model_discov = load_object(Path("outputs/gmm_clustering/model_discov.pkl"))
model_valid = load_object(Path("outputs/gmm_clustering/model_valid.pkl"))

# Set outputs folder
output_folder = Path("outputs/gmm_clustering")
output_folder.mkdir(exist_ok=True)
```

```python
# Make plots of mixture weights
plot_discov_weights = make_plot_mixture_weights(model_discov)
plot_valid_weights = make_plot_mixture_weights(model_valid)

# Combine mixture weights plots
plot_all_weights = (
    (plot_discov_weights + p9.labs(title="Discovery")) / (plot_valid_weights + p9.labs(title="Validation"))
) & p9.theme(figure_size=(5, 5))
plot_all_weights.show()

# Save plots
# save_plot(plot_discov_weights, output_folder / "plot_discov_weights.png", dpi=150)
# save_plot(plot_valid_weights, output_folder / "plot_valid_weights.png", dpi=150)

save_plot(plot_all_weights, output_folder / "plot_all_weights.png", dpi=150)
```
