# Notebook for descriptive analysis of the dataset

```python
import os

import matplotlib.pyplot as plt
import pandas as pd

from utils import plot_corr

# Hardcoded paths, change to apply to new data
data_folder = "data/readin"
preprocessed_folder = "data/preprocessed"
outputs_folder = "outputs"

data_filename = "esa_2018.csv"
vars_info_filename = "esa_grouped_vars_info.csv"

data_path = os.path.join(data_folder, data_filename)
vars_info_path = os.path.join(preprocessed_folder, vars_info_filename)

# Load data and information about variables
data = pd.read_csv(data_path, index_col=0, low_memory=False)
vars_info = pd.read_csv(vars_info_path, index_col=0, low_memory=False)
```

## Dataset

```python
print(f"Data-frame shape: {data.shape}")
print("Variable names: " + ", ".join(data.columns))
data
```

## Variables

```python
# Print information about variables
vars_info_txt_path = os.path.join(outputs_folder, "esa_vars_info.txt")
with open(vars_info_txt_path, "r", encoding="utf-8") as f:
    print(f.read())
```

```python
# Show variable grouping
vars_info.drop(columns="info").sort_values("group")
```

```python
# Variable descriptions by group
print_desc = True  # Whether to print the descriptions instead of the names
for group in vars_info["group"].unique():
    to_print = "desc" if print_desc else "var"
    print(f"{group} variables:  " + " ● ".join(vars_info.query(f"group == '{group}'")[to_print]))
```

```python
# Search for variables by parts of their name
print_desc = True  # Whether to print the descriptions instead of the names


def search_variables(search_term: str, vars_info: pd.DataFrame) -> list[tuple[str, str]]:
    matches = [
        (var, desc)
        for var, desc, *_ in vars_info.itertuples(index=False, name=None)
        if search_term.lower() in desc.lower()
    ]
    return matches


search_term = "Personen"
var_matches = search_variables(search_term, vars_info=vars_info)

to_print = 1 if print_desc else 0
print(f"Variables matching '{search_term}':  " + " ● ".join(match[to_print] for match in var_matches))
```

```python
# Variables with only missing values
print_desc = False  # Whether to print the descriptions instead of the names

to_print = "desc" if print_desc else "var"
missing_vars = vars_info.reset_index().query("type == 'all_na'")[to_print]
print(f"{len(missing_vars)} variables with only missing values: " + " ● ".join(missing_vars))
```

## Plot correlation matrices

```python
group = "socioecon"
plot_corr(
    data,
    var_names=[var for var in data.columns if var in vars_info.query(f"group == '{group}'").index.values],
    tick_label_size=13,
    title_size=20,
    title="Socioeconomic variables",
    figsize=(7, 6),
)
plt.show()
```
